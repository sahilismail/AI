```python
print("SMART STUDY PLANNER")

subjects = []
difficulties = []

count = int(input("How many subjects do you have? "))

for i in range(count):
    print("\nSubject", i + 1)

    name = input("Enter subject name: ")
    difficulty = input("Difficulty (easy/medium/hard): ").lower()

    subjects.append(name)
    difficulties.append(difficulty)


hours = float(input("\nHow many hours can you study today? "))

weights = []

for difficulty in difficulties:
    if difficulty == "hard":
        weights.append(3)

    elif difficulty == "medium":
        weights.append(2)

    else:
        weights.append(1)


total_weight = sum(weights)

print("\n----------------------------")
print("YOUR STUDY PLAN")
print("----------------------------")

for i in range(count):
    study_time = (weights[i] / total_weight) * hours

    print(
        subjects[i],
        "-",
        round(study_time, 1),
        "hours"
    )


print("\nAgent Recommendation")

hard_subjects = []

for i in range(count):
    if difficulties[i] == "hard":
        hard_subjects.append(subjects[i])


if len(hard_subjects) > 0:
    print("Give extra attention to:")

    for subject in hard_subjects:
        print("-", subject)

else:
    print("Your subjects have manageable difficulty.")


print("\nStudy plan created successfully.")
```
